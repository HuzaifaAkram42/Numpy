### 2. np.append

The numpy.append() appends values along the mentioned axis at the end of the array

https://numpy.org/doc/stable/reference/generated/numpy.append.html

```python
np.append(a,200)
```

```python
np.append(a,[200,300,500])
```

```python
b
```

```python
# append random numbers column at the end of an array
np.append(b,np.round((np.random.random((b.shape[0],1))*100)),axis=1)
```

```python
# append random numbers row at the end of an array
np.append(b,np.round((np.random.random((1,b.shape[1]))*100)),axis=0)
```

```python
# append ones numbers row at the end of an array
np.append(b,np.ones((1,b.shape[1])),axis=0)
```

```python
# append ones numbers column at the end of an array
np.append(b,np.ones((b.shape[0],1)),axis=1)
```

### 6. np.where

The numpy.where() function returns the indices of elements in an input array where the given condition is satisfied.

https://numpy.org/doc/stable/reference/generated/numpy.where.html

```python
a
```

```python
# find all indices with value greater than 50 (return only true value index)
np.where(a>50)
```

```python
# where(condition,True Value, False Value)
np.where(a>20,1,a)
```

```python
# replace all values > 50 with 0
np.where(a>50,0,a)
```

```python
np.where(a%2 == 0,0,a)
```

```python
np.where(b%5==0,5,b)
```

```python
np.where(b%7==0,0,1)
```

```python
np.where(b%7==0,"Good","Bad")
```

```python
# where does not support axis arguments
np.where(b%7==0,0,1,axis=1)
```
